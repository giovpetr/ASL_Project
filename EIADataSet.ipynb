Example pour importer data set depuis EIA https://www.eia.gov/opendata/ en utilisant la API fourni par eux

```python
import requests
import pandas as pd
from google.colab import userdata


API_KEY = userdata.get("EIA_API_KEY") # il faut definir ta clé parmi les Google Secrets

url = "https://api.eia.gov/v2/ieo/2023/data/"

params = {
    "api_key": API_KEY,
    "frequency": "annual",
    "data": [
        "value"
    ],
    "facets": {},
    "start": "2020",
    "end": "2026",
    "sort[0][column]": "period",
    "sort[0][direction]": "desc",
    "offset": 0,
    "length": 5000
}

response = requests.get(url, params=params)

data = response.json()

#print(data)

df = pd.DataFrame(data["response"]["data"])

df.head()

print("Observations:", df.shape[0])
print("Features:", df.shape[1])

print(df.columns)
```
